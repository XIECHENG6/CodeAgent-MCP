```python
# Step 2: Set up token
# Option A: Use Colab secrets
from google.colab import userdata
HF_TOKEN = userdata.get('HF_TOKEN')

# Option B: Paste directly (uncomment below)
# HF_TOKEN = "hf_xxxxx"
```

```python
# Step 3: Sync project from Drive
import os
from google.colab import drive
drive.mount('/content/drive')

# Adjust path to match your Drive folder
DRIVE_PROJECT = '/content/drive/MyDrive/CodeAgent-MCP'
LOCAL_PROJECT = '/content/CodeAgent-MCP'

!rm -rf {LOCAL_PROJECT}
!cp -r {DRIVE_PROJECT} {LOCAL_PROJECT}
os.chdir(LOCAL_PROJECT)
print(f"Working dir: {os.getcwd()}")
!ls
```

```python
# Step 4: Deploy to HF Spaces
!python deploy/upload_to_hf_space.py --token {HF_TOKEN}
```

```python
# Step 5: Verify Space is live
from huggingface_hub import HfApi
_api = HfApi(token=HF_TOKEN)
_username = _api.whoami()["name"]
print("\n" + "="*60)
print("Deployment complete!")
print(f"Space URL: https://huggingface.co/spaces/{_username}/CodeAgent-MCP")
print("="*60)
print("\nNote: First build may take 2-3 minutes.")
print("The demo requires a DeepSeek API key to use.")
```
