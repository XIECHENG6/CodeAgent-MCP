```python
# Cell 6: 测试 file_server
import asyncio
from mcp.client.stdio import stdio_client, StdioServerParameters
from mcp import ClientSession

async def test_file_server():
    params = StdioServerParameters(command="python", args=["-m", "src.mcp.servers.file_server"])
    async with stdio_client(params) as (read, write):
        async with ClientSession(read, write) as session:
            await session.initialize()
            tools = await session.list_tools()
            print(f"✅ File Server: {len(tools.tools)} tools")
            for t in tools.tools:
                print(f"   - {t.name}: {t.description}")

await test_file_server()
```

```python
# Cell 6: 测试 MCP file_server（通过独立脚本运行）
  %%writefile /tmp/test_mcp.py
  import asyncio
  import os
  os.chdir("/content/drive/MyDrive/CodeAgent-MCP")

  from mcp.client.stdio import stdio_client, StdioServerParameters
  from mcp import ClientSession

  async def test_file_server():
      params = StdioServerParameters(command="python", args=["-m", "src.mcp.servers.file_server"])
      async with stdio_client(params) as (read, write):
          async with ClientSession(read, write) as session:
              await session.initialize()
              tools = await session.list_tools()
              print(f"✅ File Server: {len(tools.tools)} tools")
              for t in tools.tools:
                  print(f"   - {t.name}: {t.description}")

              # 测试 file_list 工具
              result = await session.call_tool("file_list", {"directory": ".", "max_depth": 1})
              print(f"\n✅ file_list 调用成功:")
              print(result.content[0].text[:500])

  asyncio.run(test_file_server())
```
